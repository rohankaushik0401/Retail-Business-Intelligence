```python
import pandas as pd
import numpy as np
customers = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_customers_dataset.csv")

orders = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_orders_dataset.csv")

order_items = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_order_items_dataset.csv")

payments = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_order_payments_dataset.csv")

products = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_products_dataset.csv")

reviews = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_order_reviews_dataset.csv")

sellers = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_sellers_dataset.csv")

translation = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\product_category_name_translation.csv")

geolocation = pd.read_csv(r"C:\Users\RohanS\Documents\data retail\olist_geolocation_dataset.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "products": products,
    "reviews": reviews,
    "sellers": sellers,
    "translation": translation,
    "geolocation": geolocation
}

for name, df in datasets.items():
    print("="*50)
    print(name.upper())
    print(df.shape)
    print(df.columns.tolist())
```

```python
fact_sales = orders.merge(
    customers,
    on="customer_id",
    how="left"
)

fact_sales = fact_sales.merge(
    order_items,
    on="order_id",
    how="left"
)

fact_sales = fact_sales.merge(
    products,
    on="product_id",
    how="left"
)

fact_sales = fact_sales.merge(
    translation,
    on="product_category_name",
    how="left"
)

fact_sales = fact_sales.merge(
    payments,
    on="order_id",
    how="left"
)

fact_sales = fact_sales.merge(
    reviews,
    on="order_id",
    how="left"
)

fact_sales = fact_sales.merge(
    sellers,
    on="seller_id",
    how="left"
)

print(fact_sales.shape)
fact_sales.to_csv(
    r"C:\Users\RohanS\Documents\data retail\fact_sales.csv",
    index=False
)
```

```python

fact_sales['order_purchase_timestamp'] = pd.to_datetime(fact_sales['order_purchase_timestamp'])

fact_sales['order_delivered_customer_date'] = pd.to_datetime(fact_sales['order_delivered_customer_date'])

fact_sales['order_estimated_delivery_date'] = pd.to_datetime(fact_sales['order_estimated_delivery_date'])

fact_sales['order_approved_at'] = pd.to_datetime(fact_sales['order_approved_at'])


```

```python

fact_sales_fe=fact_sales.copy()
```

```python
fact_sales_fe["approval_time"]=fact_sales_fe['order_approved_at']-fact_sales_fe['order_purchase_timestamp']
```

```python
fact_sales_fe["delivery_days"]=fact_sales_fe['order_delivered_customer_date']-fact_sales_fe['order_purchase_timestamp']
```

```python
fact_sales_fe["delivery_delay"]=fact_sales_fe['order_estimated_delivery_date']-fact_sales_fe['order_delivered_customer_date']
```

```python
fact_sales_fe['product_volume_cm3']=fact_sales_fe['product_length_cm']*fact_sales_fe["product_height_cm"]*fact_sales_fe["product_width_cm"]
```

```python
fact_sales_fe['freight_%']=fact_sales_fe["freight_value"]*100/fact_sales_fe['price']
```

```python




fact_sales_fe['order_month']=fact_sales_fe['order_purchase_timestamp'].dt.strftime("%B")
fact_sales_fe['order_yaer']=fact_sales_fe['order_purchase_timestamp'].dt.year
fact_sales_fe['order_day']=fact_sales_fe['order_purchase_timestamp'].dt.day_name()
fact_sales_fe['order_hour']=fact_sales_fe['order_purchase_timestamp'].dt.hour
print(fact_sales.shape)
fact_sales_fe.to_csv(
    r"C:\Users\RohanS\Documents\data retail processed\fact_sales_fe.csv",
    index=False)
fact_sales_fe
```

```python
product_mart = (
    fact_sales_fe
    .groupby("product_id")
    .agg(
        category=("product_category_name_english","first"),
        total_revenue=("price","sum"),
        units_sold=("order_item_id","count"),
        avg_price=("price","mean"),
        avg_review=("review_score","mean"),
        avg_freight=("freight_value","mean"),
        total_orders=("order_id","nunique")
    )
    .reset_index()
)
print(fact_sales.shape)
product_mart.to_csv(
    r"C:\Users\RohanS\Documents\data retail processed\product_mart.csv",
    index=False)
```

```python
customer_mart = (
    fact_sales_fe
    .groupby("customer_unique_id")
    .agg(
        total_orders=("order_id","nunique"),
        total_spent=("payment_value","sum"),
        avg_review=("review_score","mean"),
        last_purchase=("order_purchase_timestamp","max")
    )
    .reset_index()
)
print(fact_sales.shape)
customer_mart.to_csv(
    r"C:\Users\RohanS\Documents\data retail processed\customer_mart.csv",
    index=False)
```

```python
seller_mart = (
    fact_sales_fe
    .groupby("seller_id")
    .agg(
        total_orders=("order_id","nunique"),
        total_revenue=("price","sum"),
        avg_delivery=("delivery_days","mean"),
        avg_review=("review_score","mean")
    )
    .reset_index()
)
print(fact_sales.shape)
seller_mart.to_csv(
    r"C:\Users\RohanS\Documents\data retail processed\seller_mart.csv",
    index=False)
```

```python
geo_mart = (
    fact_sales_fe
    .groupby("customer_state")
    .agg(
        revenue=("price","sum"),
        orders=("order_id","nunique"),
        customers=("customer_unique_id","nunique")
    )
    .reset_index()
)
print(fact_sales.shape)
geo_mart.to_csv(
    r"C:\Users\RohanS\Documents\data retail processed\geo_mart.csv",
    index=False)
```
